```python
from bs4 import BeautifulSoup
import pandas as pd
import requests

url = "https://quotes.toscrape.com/"
response = requests.get(url)

if response.status_code == 200:
  soup = BeautifulSoup(response.text, "html.parser")

  # Find all quote boxes on the page
  quotes = soup.find_all("div", class_="quote")
  quote_data = []

  for q in quotes:
    # Extract text, author, and tags
    text = q.find("span", class_="text").get_text(strip=True)
    author = q.find("small", class_="author").get_text(strip=True)

    # Some quotes have tags, let's extract them safely
    tags = [tag.get_text(strip=True) for tag in q.find_all("a", class_="tag")]

    quote_data.append(
        {"Quote": text, "Author": author, "Tags": ", ".join(tags)}
    )

  # Convert to DataFrame and save to CSV
  df = pd.DataFrame(quote_data)
  df.to_csv("quotes_data.csv", index=False)
  print("Quotes scraped and saved successfully!")
else:
  print("Failed to retrieve the webpage. Status code:", response.status_code)
```
